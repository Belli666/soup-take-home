```python
import csv, json
from pathlib import Path

root = Path('/content/soup_takehome')
print('=== verify-summary.json ===')
print((root / 'results/verify-summary.json').read_text())

print('=== quant-meta.json (summary) ===')
qm = json.loads((root / 'results/quant-meta.json').read_text())
print(json.dumps({k: v for k, v in qm.items() if not k.endswith('_sha256')}, indent=2, ensure_ascii=False))

print('=== nf4-reload-check.json ===')
for p in sorted(root.glob('logs/*-quant-*/nf4-reload-check.json')):
    print(p.parent.name, p.read_text())

print('=== nvidia-smi peaks across phases ===')
for p in sorted(root.glob('logs/*/nvidia-smi.csv')):
    rows = list(csv.DictReader(p.open()))
    if rows:
        used = [int(r[' memory.used [MiB]'].strip().split()[0]) for r in rows]
        print(f'{p.parent.name}: samples={len(used)}, min={min(used)} MiB, max={max(used)} MiB ({max(used)/1024:.3f} GiB)')
```

```python
from datetime import datetime, timezone
from pathlib import Path
import subprocess

root = Path('/content/soup_takehome')
for p in sorted(root.glob('logs/*-probes-*/raw.log')):
    lines = p.read_text(errors='replace').splitlines()
    print('=== PROBES TAIL ===')
    print('\n'.join(lines[-25:]))

for p in sorted(root.glob('logs/*-ship-*/ship-merged.json')):
    print('=== ship-merged.json ===')
    print(p.read_text())

stamp = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')
name = f'evidence-{stamp}.tar.gz'
subprocess.run([
    'tar', '-czf', name,
    '--exclude=data/raw',
    '--exclude=artifacts/adapter/checkpoint-*',
    'logs', 'results', 'data', 'soup.yaml', 'scripts', 'stages', 'run_phase.sh', 'artifacts/adapter'
], cwd=root, check=True)
sha = subprocess.check_output(['sha256sum', name], cwd=root, text=True)
(root / f'{name}.sha256').write_text(sha)
print('ARCHIVE_NAME=' + name)
print('ARCHIVE_SHA=' + sha.strip())
```
